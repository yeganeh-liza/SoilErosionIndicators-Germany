## Soil Erosion Indicators in Germany

Project Objective
This analysis compares four soil erosion indicators in Germany using QGIS and Python by calculating 
the percentage of unhealthy pixels for each indicator.

The selected indicators are:
- Harvest erosion
- Tillage erosion
- Water erosion
- Wind erosion

```python
# Create table of unhealthy (%) values by erosion/harvest indicator
import pandas as pd 
import matplotlib.pyplot as plt 
data = { "Indicator": [ "Harvest Erosion","Tillage Erosion","Water Erosion","Wind Erosion"], 
        "Unhealthy (%)":[5.0641, 25.9731, 17.5202, 1.5602]}
d_table= pd.DataFrame(data)
print(d_table)
```

```python
# Plot bar chart of unhealthy area (%) by soil erosion indicator in Germany
plt.figure(figsize=(9, 5))
plt.style.use("dark_background")
plt.grid(alpha=0.4)
plt.bar(d_table["Indicator"], d_table["Unhealthy (%)"], color="orange",width=0.6, alpha=0.8)
plt.xlabel("Erosion Indicator")
plt.ylabel("Unhealthy Area (%)")
plt.title("Unhealthy Areas Associated with Soil Erosion in Germany")
plt.tight_layout()
plt.show()
plt.show()
```
